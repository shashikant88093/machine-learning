# TensorFlow created by string , number , boolean and float  with example ?
- TensorFlow supports various data types, including strings, numbers, booleans, and floats. Here are examples of each type:
  
  1. **String**: TensorFlow can handle string data, which is often used for text processing tasks.
     ```python
     import tensorflow as tf
     string_tensor = tf.constant("Hello, TensorFlow!")
     print(string_tensor)
     ```

  2. **Number**: TensorFlow can work with integers and floating-point numbers.
     ```python
     int_tensor = tf.constant(42)
     float_tensor = tf.constant(3.14)
     print(int_tensor, float_tensor)
     ```

  3. **Boolean**: TensorFlow supports boolean values for logical operations.
     ```python
     bool_tensor = tf.constant(True)
     print(bool_tensor)
     ```

  4. **Float**: TensorFlow can handle floating-point numbers for numerical computations.
     ```python
     float_tensor = tf.constant(2.71828, dtype=tf.float32)
     print(float_tensor)
     ```

# Rank / Degree of Tensors ?

- In TensorFlow, the rank (or degree) of a tensor refers to the number of dimensions it has. Here are some examples of tensors with different ranks:
  1. **Scalar (Rank 0)**: A single value with no dimensions.
  2. **Vector (Rank 1)**: A one-dimensional array of values.
  3. **Matrix (Rank 2)**: A two-dimensional array of values.
  4. **3D Tensor (Rank 3)**: A three-dimensional array of values.
  5. **n-D Tensor (Rank n)**: A tensor with n dimensions, where n can be any positive integer.
  6. **Example of Tensors with Different Ranks**:
```python
import tensorflow as tf
# Scalar (Rank 0)
scalar_tensor = tf.constant(5)
print("Scalar Tensor (Rank 0):", scalar_tensor)
# Vector (Rank 1)
vector_tensor = tf.constant([1, 2, 3])
print("Vector Tensor (Rank 1):", vector_tensor)
# Matrix (Rank 2)
matrix_tensor = tf.constant([[1, 2], [3, 4]])
print("Matrix Tensor (Rank 2):", matrix_tensor)
# 3D Tensor (Rank 3)
three_d_tensor = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3D Tensor (Rank 3):", three_d_tensor)
# n-D Tensor (Rank n)
n_d_tensor = tf.constant([[[[1, 2], [3, 4]], [[5, 6], [7, 8]]]])
print("n-D Tensor (Rank n):", n_d_tensor)
```

## How to call rank method in tensorflow ?
- In TensorFlow, you can use the `tf.rank()` function to determine the rank (number of dimensions) of a tensor. Here's how to use it:
```python
import tensorflow as tf
# Create a tensor
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
# Get the rank of the tensor
rank = tf.rank(tensor)
print("Rank of the tensor:", rank.numpy())
```


## Shape of Tensors ?
- In TensorFlow, the shape of a tensor refers to the dimensions of the tensor. It describes how many elements are present in each dimension. The shape is represented as a tuple or a list, where each element corresponds to the size of that dimension.
- You can access the shape of a tensor using the `.shape` attribute or the `tf

.shape()` function. Here's how to do it:
```python
import tensorflow as tf
# Create a tensor
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
# Get the shape of the tensor
shape = tensor.shape
print("Shape of the tensor:", shape)
# Alternatively, using tf.shape()
shape_tensor = tf.shape(tensor)
