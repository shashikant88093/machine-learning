# Tensors
- Tensors are the fundamental data structure in TensorFlow, representing multi-dimensional arrays.
- They can be thought of as generalizations of vectors and matrices to higher dimensions.
- Tensors can have different ranks, which indicate the number of dimensions they have:
  - Rank 0: Scalar (a single value)
  - Rank 1: Vector (a one-dimensional array)
  - Rank 2: Matrix (a two-dimensional array)
  - Rank n: n-dimensional array

# TensorFlow
- TensorFlow is an open-source machine learning framework developed by Google.
- It provides a comprehensive ecosystem for building and deploying machine learning models.
- TensorFlow supports various machine learning tasks, including deep learning, reinforcement learning, and more.
- It allows developers to create complex neural networks and perform computations efficiently using data flow graphs.

## Example Code
```python
import tensorflow as tf
# Create a constant tensor
