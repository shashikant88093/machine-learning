# What is Linear Regression ?
- Linear regression is a statistical method used to model the relationship between a dependent variable and one or more independent variables. It assumes that there is a linear relationship between the variables, which can be represented by a straight line in a two-dimensional space.
- The goal of linear regression is to find the best-fitting line that minimizes the difference between the observed values and the predicted values. This is typically done by minimizing the sum of the squared differences between the observed and predicted values, known as the residuals.

## How does Linear Regression work  with TensorFlow ?
- TensorFlow provides a high-level API for building and training linear regression models. The process typically involves the following steps:
  1. **Data Preparation**: Load and preprocess the dataset, including splitting it into training and testing sets, normalizing the data, and converting it into a suitable format for TensorFlow.
  2. **Model Definition**: Define the linear regression model using TensorFlow's Keras API. This involves specifying the input shape, the number of features, and the output layer.
  3. **Loss Function**: Choose a loss function, such as mean squared error (MSE), to measure the difference between the predicted and actual values.
  4. **Optimizer**: Select an optimizer, such as stochastic gradient descent (SGD) or Adam, to update the model's weights during training.
  5. **Training**: Train the model using the training dataset, adjusting the weights to minimize the loss function. This is done by iterating over the dataset multiple times (epochs) and updating the weights using backpropagation.
  6. **Evaluation**: Evaluate the model's performance on the testing dataset to assess its accuracy and generalization ability. 

## furmula for Linear Regression
- The formula for linear regression can be expressed as:
- \[ y = mx + b \]
- Where:
  - \( y \) is the dependent variable (the value we want to predict)
  - \( m \) is the slope of the line (the coefficient for the independent variable)
  - \( x \) is the independent variable (the input feature)
  - \( b \) is the y-intercept (the value of \( y \) when \( x = 0 \))  
- Example:
```python   
import tensorflow as tf
import numpy as np
# Generate some sample data
np.random.seed(42)
x = np.random.rand(100, 1) * 10  # 100 random values between 0 and 10
y = 2.5 * x + np.random.randn(100, 1) * 0.5  # Linear relationship with some noise
# Define the linear regression model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(1,), activation='linear')
])
# Compile the model with mean squared error loss and Adam optimizer
model.compile(optimizer='adam', loss='mean_squared_error')
# Train the model on the sample data
model.fit(x, y, epochs=100, verbose=0)
# Make predictions on new data
new_x = np.array([[5], [7], [9]])
predictions = model.predict(new_x)
print("Predictions:", predictions)
```
