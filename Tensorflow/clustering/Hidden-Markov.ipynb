# Hidden Markov Models (HMM)

## 🔹 What is a Hidden Markov Model (HMM)?
A **Hidden Markov Model** is a **statistical model** used to represent systems that evolve over time where:

- The **system has hidden states** (not directly observable).
- You **only see observations** (emissions) that depend on those hidden states.
- The system follows the **Markov property**:
   > *The future state depends only on the present state, not on the full history.*

---

## 📌 Key Components of an HMM
An HMM is fully defined by **five elements**:

1️⃣ **States (S)**  
   - The hidden conditions of the system (e.g., *Sunny*, *Rainy*).  
   - Represented as `S = {s1, s2, …, sN}`.

2️⃣ **Observations (O)**  
   - What we can actually see (e.g., *carrying umbrella*, *not carrying umbrella*).  
   - Represented as `O = {o1, o2, …, oM}`.

3️⃣ **Transition Probabilities (A)**  
   - Probability of moving from one hidden state to another.  
   - `A = a_ij = P(q_{t+1} = s_j | q_t = s_i)`.

4️⃣ **Emission Probabilities (B)**  
   - Probability of observing a symbol given a state.  
   - `B = b_j(k) = P(o_k | q_t = s_j)`.

5️⃣ **Initial State Probabilities (π)**  
   - Probability of starting in each state.  
   - `π_i = P(q_1 = s_i)`.

---

## 📊 Intuition with a Simple Example
Imagine we want to **predict the weather** based on whether people are carrying umbrellas.

- **Hidden states:** ☀️ Sunny, 🌧 Rainy  
- **Observations:** 🌂 Umbrella, 🚶‍♂️ No umbrella

We **can’t directly observe the weather** every day (maybe you’re indoors),  
but we **can observe umbrellas**.  

👉 **HMM helps infer the hidden sequence of weather** from the umbrella sightings.

---

## 🔄 Three Core HMM Problems
HMMs are powerful because they solve **three main problems**:

1️⃣ **Evaluation**  
   - *Given an HMM and an observation sequence, what is the probability of that sequence?*  
   - ✅ **Solution:** Forward algorithm.

2️⃣ **Decoding**  
   - *Given an HMM and an observation sequence, what is the most likely sequence of hidden states?*  
   - ✅ **Solution:** Viterbi algorithm.

3️⃣ **Learning (Training)**  
   - *Given observations (and no model), how do we estimate HMM parameters (A, B, π)?*  
   - ✅ **Solution:** Baum–Welch algorithm (Expectation-Maximization).

---

## 🔹 How It’s Different from Normal Models
✅ Models **temporal dependencies** (data has order).  
✅ Works with **hidden states** (you infer what you can’t directly see).  
❌ Linear regression or logistic regression can’t do this because they assume independent data points.

---

## ✅ Real-World Applications of HMM
- 🎙 **Speech Recognition** – Convert sound waves (observations) to words (hidden states).
- 🏷 **Part-of-Speech Tagging** – Assign tags like *noun, verb* (hidden states) to words in a sentence.
- 🧬 **DNA/Genomic Sequencing** – Identify gene regions (hidden states) from DNA base sequences.
- 📈 **Finance** – Model market regimes (bull/bear states) based on stock prices.
- 🌦 **Weather Forecasting** – Infer weather patterns from indirect signals.

---

## 🔢 Quick Pseudo-Math Representation
HMM defines a **joint probability** of hidden states (Q) and observations (O):

\[
P(Q, O) = P(q_1) \prod_{t=2}^{T} P(q_t | q_{t-1}) \prod_{t=1}^{T} P(o_t | q_t)
\]

Where:
- `Q = (q1, q2, …, qT)` = hidden states over time  
- `O = (o1, o2, …, oT)` = observed sequence  

---

## 📦 Python Example (hmmlearn library)

```python
from hmmlearn import hmm
import numpy as np

# HMM with 2 hidden states
model = hmm.MultinomialHMM(n_components=2)

# Observations: Umbrella(1) or No umbrella(0)
observations = np.array([[1], [1], [0], [1], [0]])

# Fit model (Baum-Welch)
model.fit(observations)

# Decode sequence (Viterbi)
logprob, hidden_states = model.decode(observations, algorithm="viterbi")
print("Most likely hidden states:", hidden_states)
