# What is clustering ?
- Clustering groups the weights of a neural network into a set number of shared values (clusters).

- Instead of each weight having a unique float value, weights are forced to share the same value from a limited set.

- For example:

- Normally: 10,000 weights might have 10,000 unique float values.

- After clustering (e.g., 32 clusters): all weights are mapped to only 32 distinct values.

- ✅ Why do this?

- Smaller model size → easier to store and deploy (especially on mobile & edge devices).

- Better compression → weights become more uniform, leading to better gzip/zip compression.

Can be combined with quantization for even greater efficiency.

🔹 How does it work in TensorFlow?
- Uses the tensorflow_model_optimization API.

- You wrap your model or specific layers with cluster_weights().

- Train (or fine-tune) the clustered model to maintain accuracy.

```python

import tensorflow as tf
import tensorflow_model_optimization as tfmot

# Define a simple model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(128, activation='relu', input_shape=(784,)),
    tf.keras.layers.Dense(10, activation='softmax')
])

# Apply clustering
clustering_params = {
    'number_of_clusters': 16,
    'cluster_centroids_init': tfmot.clustering.keras.CentroidInitialization.LINEAR
}

clustered_model = tfmot.clustering.keras.cluster_weights(model, **clustering_params)

# Compile and train (fine-tuning)
clustered_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
# clustered_model.fit(x_train, y_train, epochs=1)

# Strip clustering to get a deployable model
final_model = tfmot.clustering.keras.strip_clustering(clustered_model)
```

# What is K-means ?
- K-means is one of the most popular unsupervised machine learning algorithms used for clustering.
- Instead of predicting labels (like in supervised learning), K-means groups similar data points into clusters based on their features.
🔹 How K-means Works (Step by Step)
1️⃣ Choose the number of clusters (K)

You decide how many groups you want the algorithm to find.

2️⃣ Initialize centroids

Randomly select K points from the dataset as the initial “centers” of clusters (centroids).

3️⃣ Assign each data point to the nearest centroid

Use a distance metric (usually Euclidean distance) to find which centroid is closest to each point.

4️⃣ Recalculate centroids

Take the mean of all points assigned to each cluster → this becomes the new centroid.

5️⃣ Repeat steps 3 and 4 until:

Centroids don’t move much anymore (convergence), OR

A maximum number of iterations is reached.
