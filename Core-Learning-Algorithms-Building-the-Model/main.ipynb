# Whats is estimator?
- In the context of machine learning, an estimator is any object that learns from data. It can be a classification, regression, or clustering algorithm, or a transformer that extracts or filters useful features from raw data.
 In classification, an estimator is specifically referred to as a classifier, which is a type of function where the response is discrete. This means that a classifier predicts the class label of a given input data point from a finite set of classes.



- For classification Task There are thousands of estimators.
- DNNClassifier (Deep Neural Network)
- LinearClassifier

Warning: Estimators are not recommended for new code. Estimators run v1.Session-style code which is more difficult to write correctly, and can behave unexpectedly, especially when combined with TF 2 code. Estimators do fall under our compatibility guarantees, but will receive no fixes other than security vulnerabilities. See the migration guide for details.
